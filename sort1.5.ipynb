# Sorting Galaxies: (Type 0 - Type 1.5)

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

Read all files

```python
catgama = pd.read_csv('InputCatGAMADR3.csv')
catclusters = pd.read_csv('samiDR3InputCatClusters.csv')
stelkin = pd.read_csv('samiDR3StelKin.csv')
vismorph = pd.read_csv('samiDR3VisualMorphology.csv')
```

```python
stelkinids = set(stelkin['catid'])
vismorphids = set(vismorph['catid'])
same = (stelkinids == vismorphids)

print(len(stelkinids))
print(len(vismorphids))
same
```

**stelkin and vismorph contain the same galaxies**


```python
print(len(stelkin))
print(len(stelkinids))

print(len(vismorph))
print(len(vismorphids))
```

Duplicated catids in stelkin to be removed before merge 
(noticed longer stelkin file but found same number of galaxies before)

```python
stelkin = stelkin.drop_duplicates(subset='catid')
```

Sort these galaxies by type 

```python
mask = (vismorph['type'] >= 0.0) & (vismorph['type'] <= 1.5)
etg = vismorph[mask]
```

note: this exclude unclassified galaxies (type < 0)

To plot FJ relation, sigma_re is required. Therefore, galaxies must appear in stelkin to be plotted. Since stelking and vismorph contain the same galaxies, sorting vismorph is sufficient for plot. 

```python
etg_sigma = pd.merge(etg, stelkin[['catid','sigma_re']], on='catid')
(etg_sigma)
```

```python
catgama_mstar = pd.merge(etg_sigma, catgama[['catid','mstar']], on='catid')
catclusters_mstar = pd.merge(etg_sigma, catclusters[['catid','mstar']], on='catid')
etg_sigma_mstar = pd.concat([catgama_mstar, catclusters_mstar])
(etg_sigma_mstar)
```

note: not all galaxies from etg_sigma have a matching mstar

```python
etg_sigma_mstar_cleaned = etg_sigma_mstar.dropna(subset=['sigma_re'])
(etg_sigma_mstar_cleaned)
```
